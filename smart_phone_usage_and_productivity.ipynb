```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace (" ", "_")
```

```python
df.columns
```

```python
df = df.rename(columns = {"level_of_student_(if_student)" : "level_of_student"})
```

```python
df.columns
```

```python
 df = df.rename(columns = {"smartphone_duration?" : "smartphone_duration"})
 df = df.rename(columns = {"smartphone_usage_select_all_that_apply" : "smartphone_usage"})
```

```python
df.columns
```

```python
efficiency_mapping = {"Not at all":"0", "Slightly" :"1", "Moderately": "2", "Very much": "3", "Extremely": "4"}
df["phone_efficiency"] = df["efficiency_of_phone"].map(efficiency_mapping)
```

```python
df[["efficiency_of_phone","phone_efficiency"]].head()
```

```python
productivity_impact_mapping = {"Strongly disagree": "1", "Disagree": "2", "Neutral": "3", "Agree": "4", "Strongly agree": "5"}
df["productivity_impact"] = df["smartphone_productivity_impact"].map(productivity_impact_mapping)
```

```python
df[["smartphone_productivity_impact" , "productivity_impact"]].head()
```

```python
df.isnull().sum()
```

```python
df["productivity_impact"] = df["productivity_impact"].fillna(
    df["productivity_impact"].mode()[0]
)
```

```python
df["productivity_impact"].dtype
```

```python
df["productivity_impact"] 
```

```python
categorical_cols = df.select_dtypes(include="str").columns

for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
numeric_cols = df.select_dtypes(include=["int64", "float64"]).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())
```
